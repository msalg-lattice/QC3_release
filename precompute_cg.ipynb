```python
import pickle
from sympy.physics.quantum.cg import CG

def precompute_cg(Jmax, Smax):
    cg_table = {}
    for J in range(Jmax + 1):
        for S in range(Smax + 1):
            for ell in range(abs(J - S), J + S + 1):
                for m in range(-S, S + 1):
                    for mJ in range(-J, J + 1):
                        for m_ell in range(-ell, ell+1):
                            val = float(CG(ell, m_ell, S, m, J, mJ).doit().n())
                            cg_table[(ell, m_ell, S, m, J, mJ)] = val
    return cg_table
```

```python
with open('base_code/Kdf3/cg_table.pkl', 'wb') as f:
    pickle.dump(precompute_cg(5, 3), f)
```
